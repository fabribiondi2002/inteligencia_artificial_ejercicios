## Clase 6 - EDA automático

### 1. Biblioteca ydata-profiling

```python
import pandas as pd
import seaborn as sns
from ydata_profiling import ProfileReport
import webbrowser
import os
from sklearn.model_selection import train_test_split
```

```python
# Cargar el dataset de Titanic
df = sns.load_dataset("titanic")
df.head()
```

### 1. EDA automático con biblioteca ydata-profiling

```python
# Creamos el reporte
# minimal = se puede poner en True para que el reporte salga más rápido pero con menos detalles
# explorative = muestra todos los detalles (visualizaciones adicionales y otra info)
# sensitive = no muestra valores, muestra solo las estadísticas
profile_1 = ProfileReport(df, title="Pandas Profiling Report", minimal=False, 
                          explorative=True, sensitive=False, 
                          correlations={
                              "auto":     {"calculate": False},
                              "pearson": {"calculate": True}, 
                              "spearman": {"calculate": True}, 
                              "kendall": {"calculate": True}, })
```

```python
# Exportar el repote a un archivo HTML o Json
profile_1.to_file("reportes/titanic_report_1.html")

profile_1.to_file("reportes/titanic_report_1.json") # Esta opción se usa para customizar el reporte 
```

```python
# Pequena función para abrir el reporte en el navegador
def open_html_report(report_path):
    file_path = os.path.abspath(report_path)

    # Abrir el reporte en el browser
    webbrowser.open(f"file:///{file_path}")

    return
```

```python
# Si lo queremos mostrar directamente en la notebook, correr la siguiente línea:
# profile.to_notebook_iframe() 
```

```python
# Abrir el reporte 1 en el browser
open_html_report("reportes/titanic_report_1.html")
```

### 2. EDA automático con biblioteca sweetviz

```python
import sweetviz as sv
```

```python
sv.analyze(df).show_html("reportes/sweetviz_report.html")
```

```python
# Comparar 2 datasets (por ejemplo Train y Test)

train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

# Generar el reporte de comparación
comparacion = sv.compare([train_df, "Train"], [test_df, "Test"])

# Guardar el reporte HTML
comparacion.show_html("reportes/sweetviz_comparacion.html")
```
