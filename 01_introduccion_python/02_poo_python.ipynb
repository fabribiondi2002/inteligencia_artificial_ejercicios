## Introducción a POO

En el mundo real, la mayoría de las cosas tinen un "Plano" y múltiples **instancias** de él.

Por **Plano**, nos referimos a un conjunto abstracto de propiedades y comportamientos.

Tomemos, por ejemplo, un **auto**. Es un plano o una idea general, que cubre propiedades como tener cuatro ruedas, un color, potencia del motor, y así sucesivamente. Los autos que vemos en las calles son instancias específicas de este plano general, cada uno con sus características únicas como el color, la marca y el modelo.

¿Cuántas instancias puede tener un plano?

**Tantas como sea necesario**

En programación, existe un paradigma que sigue el mismo principio que los planos e instancias. Se llama programación orientada a objetos (OOP). En OOP, los planos se conoces como **clases** y las instancias se conocen como **objetos**.

Imaginemos una empresa realizando una encuesta por correo electrónico. Envían la misma plantilla de encuesta a todos sus clientes, quienes luego completan con su propia información.

* La planilla de encuenta: Es la **clase**
* El formulario de encuenta completado por un cliente: Es el **objeto**

En Python, podemos definir una clase utilizando la palabra clave **class** seguida del **nombre** de la clase y dos puntos.


```
class Auto:
```

image.png

En el mundo real, todo tiene características distintivas: un perro tiene su raza, color y nombre; un auto tiene su marca, modelo y color.

En la programación, las clases y los objetos reflejan este concepto con **atributos**. Los atributos son las propiedades que definen la individualidad de un objeto dentro de una clase.

Para agregar atributos a una clase, debemos definir el método:
```
__init__
```
El primer parámetro de este método siempre es **self**, que representa la instancia de la clase. Después de self, especificamos los atributos que deseamos incluir.

Luego, dentro de la función, asignamos valores a los atributos del objeto inicializado, estableciendo su estado inicial.

````python
class Auto:
  # Inicialización de atributos
  def __init__(self, marca, modelo, color):
    # Asignación de valores a atributos
    self.marca = marca
    self.modelo = modelo
    self.color = color

# Creación de un objeto de la clase Auto
mi_auto = Auto("Toyota", "Corolla", "Rojo")

print(mi_auto)
````

Después de que se crea un objeto, podemos acceder a sus atributos utilizando la notación de punto, con la variable que contiene el objeto.

````python
print(mi_auto.color)
print(mi_auto.marca)
print(mi_auto.modelo)
````

Además de los atributos podemos agregar **comportamientos** personalizados a una clase definiendo **funciones** dentro de ella.

Estas funciones, conocidas como **métodos**, deben incluir el parámetro **self** para interactuar con la instancia de la clase.

Podemos llamar a estos métodos utilizando la notación de punto, similar a como accedemos a los atributos.

````python
class Perro:
  # Inicialización de atributos
  def __init__(self, nombre, raza, edad):
    # Asignación de valores a atributos
    self.nombre = nombre
    self.raza = raza
    self.edad = edad

  def ladrar(self):
    print("Guau!")

# Creación de un objeto de la clase Perro
mi_perro = Perro("Fido", "Labrador", 3)

# Llamada al método ladrar
mi_perro.ladrar()
````

La principal diferencia entre las **funciones** y los **métodos** es que las funciones son independientes y pueden ser llamadas por sí mismas, mientras que los métodos están asociados con una clase y sólo pueden ser llamados con su instancia.

Esto significa que no podemos llamar a un método sin tener la instancia de una clase donde ese método está definido.


```
print(): # es una función
mi_perro.ladrar(): # es un método
```

## Herencia

La fortaleza de OOP reside en organizar un programa de manera que sus diversos componentes, tratados como clases y objetos, puedan interactuar sin problemas.

¿Qué es una clase?

**Un plano para crear objetos, definir sus características y comportamiento.**

La **herencia** es un concepto clave para situaciones en las que tenemos una clase existente con atributos y comportamientos definidos, y necesitamos una nueva clase que no solo comparta estas características sino que también tenga las suyas propias.

La herencia permite a la nueva clase "heredar" propiedades de la clase existente mientras añade o modifica características según sea necesario.

````python
class Animal:
  def __init__(self, name):
    self.name = name

  def mover(self):
    print("El animal se está moviendo.")

# Herencia de la clase animal
class Perro(Animal):
  # Comportamiento especifico:
  def ladrar(self):
    print("Guau!")

# Creación de un objeto de la clase Perro
mi_perro = Perro("Firulais")

print(mi_perro.name)
mi_perro.mover()

# Comportamiento específico
mi_perro.ladrar()
````

Una clase de la cual otras heredan se conoce como una **superclase** o clase **padre**. Por el contrario, una clase que hereda de otra clase de llama **subclase** o clase **hija**.

En el ejemplo anterior, la clase Perro es clase hija, y la clase Animal es clase padre.

Al definir una clase hija, incluimos el nombre de la clase padre entre paréntesis.


```
class Auto(Vehiculo)
```

La clase hija hereda todos los atributos y métodos de la clase padre.


**¿Y si queremos no solo heredar atributos sino también agregar otros especificos a una clase hija?**

En este caso, definimos un método __init__ en la clase hija. Usamos:

```
super().__init__()
```
para heredar atributos de la clase padre, y luego definimos cualquier atributo adicional como de costumbre.


````python
# clase padre
class Animal:
  def __init__(self, nombre):
    self.nombre = nombre

  def mover(self):
    print("El animal se está moviendo")

# clase hija
class Perro(Animal):
  def __init__(self, nombre, raza, edad):
    # Inicializamos atributo de la superclase (padre)
    super().__init__(nombre)
    # agregamos los atributos especificos para perro
    self.raza = raza
    self.edad = edad

  def ladrar(self):
    print("Guau!")

# creamos un objeto de la clase Perro
mi_perro = Perro("Thanos", "Dogo", 5)

print(mi_perro.nombre)
print(mi_perro.raza)
print(mi_perro.edad)
````

La función **super()** se utiliza para referirse a la clase padre, permitiendo el acceso a sus métodos y atributos desde dentro de una subclase (hija).

Ejemplo:
La clase **Estudiante** hereda de la clase **Persona**. La clase Persona tiene los atributos nombre y edad.
Definimos la clase Estudiante y añadimos un atributo de facultad en ella.

````python
class Persona:
  def __init__(self, nombre, edad):
    self.nombre = nombre
    self.edad = edad

class Estudiante(Persona):
  def __init__(self, nombre, edad, facultad):
    super().__init__(nombre, edad)
    self.facultad = facultad

estudiante = Estudiante("Alexis", 28, "FRC")
print(estudiante.nombre)
print(estudiante.edad)
print(estudiante.facultad)
````

Podemos definir métodos con el mismo nombre en ambas clases, padre e hijo, pero pueden realizar operaciones diferentes.
Esto se conoce como **sobrecarga de métodos**.

Por ejemplo, tenemos la clase Animal con un método sonido.
La clase hijo Perro y Gato heredan el método sonido de Animal, pero lo sobrecarga para adaptarse a sus necesidades específicas.

````python
# Clase padre
class Animal:
  def __init__(self, nombre):
    self.nombre = nombre

  # Método de sonido generico para cualquier animal
  def sonido(self):
    print("El animal hace un sonido.")

# Clase hijo
class Perro(Animal):
  def __init__(self, nombre, raza, edad):
    super().__init__(nombre)
    self.raza = raza
    self.edad = edad

  # sobrecarga del método sonido para perro
  def sonido(self):
    print("Guau!")

# Clase hijo
class Gato(Animal):
  def __init__(self, nombre, raza, edad):
    super().__init__(nombre)
    self.raza = raza
    self.edad = edad

  # sobrecarga del método sonido para gato
  def sonido(self):
    print("Miau!")

# Creación de instancias
mi_perro = Perro("Scooby", "Gran Danés", 7)
mi_gato = Gato("Campanita", "Siames", 2)

# Usando los metodos sobrecargados
mi_perro.sonido()
mi_gato.sonido()
````

Podemos usar la función **super()** si queremos llamar a un método de la clase padre mientras lo sobrecargamos.

Esto es útil cuando queremos agregar alguna funcionalidad al método de la clase hijo sin cambiar el original.

````python
# Clase padre
class Animal:
  def __init__(self, nombre):
    self.nombre = nombre

  # Método de sonido generico para cualquier animal
  def sonido(self):
    print("El animal hace un sonido.")

# Clase hijo
class Perro(Animal):
  def __init__(self, nombre, raza, edad):
    super().__init__(nombre)
    self.raza = raza
    self.edad = edad

  # Sobrecarga del método sonido para perro
  def sonido(self):
    # Llamamos al meteodo de sonido desde Animal
    super().sonido()
    print("Guau!")

# Creación de instancias
mi_perro = Perro("Scooby", "Gran Danés", 7)

# Usando los metodos sobrecargados
mi_perro.sonido()
````

La sobrecarga de métodos es una demostración de otro concepto clave en OOP **polimorfismo**.

El polimorfismo permite a los objetos utilizar métodos a su manera, incluso si comparten el mismo nombre.

En este ejemplo, aunque cada animal en la lista de animales puede ser de una subclase diferente, el código puede llamar a sonido() en cada uno sin necesidad de conocer su tipo específico:

````python
# Clase padre
class Animal:
  def __init__(self, nombre):
    self.nombre = nombre

  # Método de sonido generico para cualquier animal
  def sonido(self):
    print("El animal hace un sonido.")

# Clase hijo
class Perro(Animal):
  def __init__(self, nombre, raza, edad):
    super().__init__(nombre)
    self.raza = raza
    self.edad = edad

  # Sobrecarga del método sonido para perro
  def sonido(self):
    # Llamamos al meteodo de sonido desde Animal
    super().sonido()
    print("Guau!")

# Clase hijo
class Gato(Animal):
  def __init__(self, nombre, raza, edad):
    super().__init__(nombre)
    self.raza = raza
    self.edad = edad

  # sobrecarga del método sonido para gato
  def sonido(self):
    print("Miau!")

# Creación de instancias
mi_perro = Perro("Scooby", "Gran Danés", 7)
mi_gato = Gato("Campanita", "Siames", 2)

animals = [mi_perro, mi_gato]

for animal in animals:
  animal.sonido()
````

Los 4 principios principales de POO, son:
*   Polimorfismo
*   Abstracción
*   Herencia
*   Encapsulación

## Encapsulamiento

El encapsulamiento de datos es una idea clave para hacer código con objetos (como en juegos o aplicaciones) más seguro y limpio.

Esto significa mantener algunas partes de un objeto privado para que sólo ciertas partes del código puedan ser cambiadas. Ayuda a prevenir error y mantener el código fácil de gestionar.

Consideramos la clase **Auto** proporcionada a continuación.

````python
class Auto:
  def __init__(self, modelo, anio, odometro):
    self.modelo = modelo
    self.anio = anio
    self.odometro = odometro

  def describir_auto(self):
    print(self.anio, self.modelo)

  def leer_odometro(self):
    print("Odometro: ", self.odometro, " km")

mi_auto = Auto("Toyota", 2015, 150000)
mi_auto.describir_auto()
mi_auto.leer_odometro()
````

````python
mi_auto.odometro = 80000
mi_auto.leer_odometro()
````

En programación, a veces es crucial **proteger** ciertos atributos y métodos de clase para que no se accedan desde fuera de la clase. Esto se llama **Encapsulamiento de datos** y garantiza la integridad y seguridad de los datos, evitando modificaciones no deseadas o dañinas.

En Python, el encapsulamiento de datos tiene dos niveles. El primero involucra poner un prefijo a un atributo con un **guión bajo simple** _ señalando que está destinado para uso interno y debe ser visto como **pretegido**

Haciendo este cambien en el atributo del **odómetro** tenemos:

````python
class Auto:
  def __init__(self, modelo, anio, odometro):
    self.modelo = modelo
    self.anio = anio
    self._odometro = odometro # ahora esta protegido

  def describir_auto(self):
    print(self.anio, self.modelo)

  def leer_odometro(self):
    print("Odometro: ", self._odometro, " km")

mi_auto = Auto("Toyota", 2015, 150000)
mi_auto.describir_auto()
mi_auto.leer_odometro()
````

````python
mi_auto._odometro = 80000
mi_auto.leer_odometro()
````

Los atributos con un solo guión bajo son accesibles pero se consideran protegidos por convención, señalando que son para uso interno y se deben acceder con precaución fuera de la clase.

El siguiente nivel de encapsulamiento de datos implica hacer un atributo **privado**.

Esto se logra prefijando el nombre del atributo con **dos guiones bajos**.

En este caso, a diferencia de los atributos protegidos, es no es solo una convención sino que limita su acceso fuera de la clase a través del enmascaramiento de nombres, mejorando la protección de datos y la **encapsulación**. Este método se utiliza para datos sensibles o internos, desalentando fuertemente el acceso externo.

````python
class Auto:
  def __init__(self, modelo, anio, odometro):
    self.modelo = modelo
    self.anio = anio
    self.__odometro = odometro # ahora esta privado

  def describir_auto(self):
    print(self.anio, self.modelo)

  def leer_odometro(self):
    print("Odometro: ", self.__odometro, " km")

mi_auto = Auto("Toyota", 2015, 150000)
mi_auto.describir_auto()
mi_auto.leer_odometro()
````

````python
mi_auto.__odometro = 80000
mi_auto.leer_odometro()
````

Acceder a un atributo privado con doble guion bajo desde fuera de la clase provoca un error, pero es accesible dentro de los métodos de la clase. Esto demuestra la **encapsulación**, protegiendo los datos sensibles del acceso externo y asegurando que sólo sean accesibles a través de métodos específicos, alineándose con los principios de la programación orientada a objetos.

Acceder a un atributo privado directamente desde fuera de su clase generalmente se desaconseja en Python. Sin embargo, Python emplea el **name mangling** para atributos privados, lo que signifca que podemos acceder a ellos utilizando una convención de nombres específicios desde fuera de la clasi si es necesario.

````python
class Auto:
  def __init__(self, modelo, anio, odometro):
    self.modelo = modelo
    self.anio = anio
    self.__odometro = odometro # privado
  def describir_auto(self):
    print(self.anio, self.modelo)

  def leer_odometro(self):
    print("Odometro: ", self.__odometro, " km")

mi_auto = Auto("Toyota", 2015, 150000)

# accediendo usando name mangling
print(mi_auto._Auto__odometro)
````

Sin embargo, este enfoque debería usarse con moderación, ya que elude los principios de encapsulación previstos al hacer el atributo privado.

También podemos designar métodos como protegidos o privados, siguiendo la misma convención que con los atributos.

Los métodos protegidos están prefijados con un único guión bajo y pueden ser accedidos dentro de la clase y sus subclases.

Sin embargo, lo métodos privados, marcados porun doble guión bajo, no pueden ser accedidos directamente desde fuera de la clase.

````python
class Auto:
  def __init__(self, modelo, anio, odometro):
    self.modelo = modelo
    self.anio = anio
    self.__odometro = odometro

  def _describir_auto(self): # metodo protegido
    print(self.anio, self.modelo)

  def __leer_odometro(self): # metodo privado
    print("Odometro: ", self.__odometro, " km")

mi_auto = Auto("Toyota", 2015, 150000)

# llamando al metodo protegido
mi_auto._describir_auto()

# llamando al metodo privado
mi_auto.__leer_odometro()
````

*   *Usando un sólo guión bajo (_), indicamos que los atributos o métodos son protegidos, destinados para uso interno.*
*   *Los guiones bajos dobles (__) hacen que los atributos o métodos sean privados, restringiendo el acceso a la misma clase, lo que fortalece a la seguridad de los datos.*
*   *Estas convenciones ayudan a gestionar el acceso y mantener una estructura clara en tu código.*

## Métodos de Clase y Estáticos

Los métodos de los objetos que hemos visto hasta ahora son llamados por una instancia de una clase. Sin embargo, hay otros tipos de métodos que una clase puede tener:
* Métodos de clase
* Métodos estáticos

Los **métodos de clase** se llaman en la **clase** misma, no en instancias individuales. Esto permite su uso sin necesidad de crear una instancia de clase.

Son especialmente útiles para acciones relevantes para la clase en su conjunto, más que acciones limitadas a un solo objeto.

````python
class Libro:
  def __init__(self, title, author):
    self.title = title
    self.author = author

  # método regular
  def describir_libro(self):
    print(self.title, 'by', self.author)

  # método de clase
  @classmethod
  def libros_en_serie(cls, nombre_serie, numero_de_libros):
    print(f'Hay {numero_de_libros} libros en la serie {nombre_serie}.')

# creación de una instancia de Libro
mi_libro = Libro("Harry Potter", "J.K. Rowling")

# Usando el metodo de instancia para describir el libro
mi_libro.describir_libro()

# Usando el método de clase para mostrar informacion acerca de la serie
Libro.libros_en_serie("Harry Potter", 7)
````

Los métodos de clase se crean utilizando el decorador @classmethod y toman el argumento **cls**, que se refiere a la **class** en sí misma.

Para llamar a un método de clase no necesitamos crear una instancia de la clase. En su lugar, simplemente se usa el nombre de la clase, seguido de un punto y el nombre del método de la clase.

Los métodos de clase son muy útiles para acceder o gestionar datos y funcionalidad que se relacionan con la clase en sí, en lugar de cualquier instancia específica.

Las instancias comparten todo lo que una clase tiene, incluyendo los métodos de clase. Esto significa que también se puede llamar a un método de clase en las instancias.

````python
class Libro:
  def __init__(self, title, author):
    self.title = title
    self.author = author

  # método regular
  def describir_libro(self):
    print(self.title, 'by', self.author)

  # método de clase
  @classmethod
  def libros_en_serie(cls, nombre_serie, numero_de_libros):
    print(f'Hay {numero_de_libros} libros en la serie {nombre_serie}.')

# creación de una instancia de Libro
mi_libro = Libro("Harry Potter", "J.K. Rowling")

# Usando el metodo de instancia para describir el libro
mi_libro.describir_libro()

# Llamando al método de clase con la instancia
mi_libro.libros_en_serie("Harry Potter", 7)
````

Los **métodos estáticos** son similares a los métodos de clase excepto que no reciben argumento adicionales, son idénticos a las funciones normales que pertenecen a una clase.

Se marcan con el decorador **@staticmethod**

````python
class Libro:
  def __init__(self, title, author):
    self.title = title
    self.author = author

  # método regular
  def describir_libro(self):
    print(self.title, 'by', self.author)

  # método estáticos
  @staticmethod
  def libros_en_serie(nombre_serie, numero_de_libros): # no tiene el argumento cls
    print(f'Hay {numero_de_libros} libros en la serie {nombre_serie}.')

# creación de una instancia de Libro
mi_libro = Libro("Harry Potter", "J.K. Rowling")

# Usando el metodo de instancia para describir el libro
mi_libro.describir_libro()

# Llamando al método estático
Libro.libros_en_serie("Harry Potter", 7)
````

**¿Cuándo deberíamos usar métodos estáticos en lugar de métodos de clase?**

Los métodos estáticos no aceptan el parámetro cls, lo que significa que no pueden acceder o modificar el estado de la clase.

Son útiles cuando necesitamos una funcionalidad que no depende del comportamiento de la clase o del estado de la instancia y no lo afecta.

Esencialmente, los métodos estáticos son adecuados para tareas que sono autónomas y no requieren conocimiento de la clase o de la instancia

Un método estático:
* No necesita ni self (instancia), ni cls (clase).
* Se comporta como una función normal que está dentro de la clase solo por organización o contexto.
* Se puede llamar desde la clase o desde la instancia.

````python
class Calculadora:
    @staticmethod
    def sumar(a, b):
        return a + b

    @staticmethod
    def restar(a, b):
        return a - b

print(Calculadora.sumar(10, 5))
print(Calculadora.restar(10, 5))


c = Calculadora()
print(c.sumar(3, 2))
````
