***Casos controversiales en conversiones de datos:***

*Ariane 5: Falla en el software. 1996. Error de conversión de punto flotante de 64 bits a entero de 16 bits, overflow aritmético.*

[Video](https://www.youtube.com/watch?v=EbzmL-fYuAM&ab_channel=DRAKOTAKOMAX)

El Ariane 5 falló el 4 de junio de 1996 en su vuelo inaugural. A los 37 segundos del despegue, el cohete se desvió de su trayectoria y explotó en el aire.

💥 **Causa del fallo**

La causa principal fue un error de software. Más específicamente:
*   El sistema de control de vuelo del Ariane 5 reutilizaba software del Ariane 4, que era un cohete anterior con diferentes condiciones de vuelo.
*   Ese software intentó convertir un número de coma flotante (float) a un entero (int), pero el valor era demasiado grande para caber en el tipo de dato entero.
*   Esto causó una excepción de overflow (desbordamiento).
*   El sistema de control no manejó correctamente la excepción, y como resultado, falló completamente.
*   El sistema de respaldo, que debía tomar el control, tenía el mismo software con el mismo error, así que también falló.
*   Sin sistema de guía, el cohete se desvió y se autodestruyó automáticamente por seguridad.
*   El software del Ariane 5 estaba escrito en Ada, un lenguaje de programación diseñado específicamente para sistemas embebidos y de tiempo real, como los que se usan en aeronáutica, defensa y control de misiones espaciales.

👨‍🔬 **¿Por qué pasó esto?**

El software venía del Ariane 4, donde ese tipo de conversión era segura, pero en el Ariane 5 los valores eran más grandes debido a su mayor potencia y velocidad.

No se desactivó esa parte del software que ya no era necesaria para el Ariane 5.

No se validaron los datos esperados ni se preparó el código para manejar errores en tiempo de ejecución.

**📚 Lección aprendida**

Este desastre es uno de los ejemplos más famosos de cómo un pequeño error de software no manejado puede causar la pérdida de millones de dólares (el cohete y la carga costaban más de $500 millones).

-------------------------

*Caso del Mars Climate Orbiter (1999). NASA perdió una sonda espacial de $327 millones, por un error de conversión de unidades en el software.*

[Video](https://www.youtube.com/watch?v=ktXCTwHT0O4&ab_channel=Katedra)

❓**¿Qué pasó?**

La sonda Mars Climate Orbiter fue lanzada por la NASA en 1998 para estudiar el clima de Marte. En septiembre de 1999, al llegar a Marte, la nave entró demasiado bajo en la atmósfera, se desintegró o se perdió para siempre.

🧨 **Causa del error**

El software desarrollado por Lockheed Martin enviaba fuerza en libras-segundo (pound-seconds).

Pero el software de la NASA esperaba fuerza en newtons-segundo (newton-seconds).

**¡No hubo conversión de unidades entre sistemas!**

Como resultado, los cálculos de trayectoria estaban mal y la nave terminó entrando en la atmósfera 100 km más abajo de lo previsto.

🔢 **Conversión incorrecta**

1 libra-fuerza (lbf) = 4.45 newtons

--------------------

*Glitch de Therac-25 (años 80, pero relevante hasta hoy)*

Aunque es de los 80, este caso sigue siendo el manual de lo que NO debe hacerse en software crítico de salud. El software del Therac-25, una máquina de radioterapia, tenía errores de concurrencia y validación que mataron a varias personas.

🧨 **¿Qué pasó?**

El Therac-25 era una máquina que usaba radiación para tratar el cáncer.

Tenía dos modos: uno de electrones (baja energía) y otro de rayos X (alta energía).

El problema principal fue un error en el software de control del sistema, específicamente una condición de carrera causada por:

Cambios rápidos de configuración por parte del operador (por ejemplo: cambiar de modo de electrones a rayos X).

El software no actualizaba correctamente el estado interno del sistema en memoria.

Como consecuencia, la máquina creía que estaba en modo seguro (baja energía), cuando en realidad estaba aplicando alta energía sin el atenuador físico.

El software no detectaba ni reportaba este estado inconsistente y continuaba operando.

⚠️ **¿Qué error de software fue?**

Mala gestión de concurrencia (race condition)

Falta de validación del estado antes de ejecutar

Mal manejo de errores: el software asumía que todo estaba bien

No existía log de fallos ni mensajes de advertencia visibles

Se confiaba solo en software, sin sensores físicos como backup

##Bucle For

La iteración se utiliza para automatizar tareas que deben realizarse una y otra vez.

La iteración hace que los programas sean más simples, más rápidos y reduce los errores.

image.png

````python
print("hola")
print("hola")
print("hola")
````

Un **bucle for** se utiliza para ejecutar la misma instrucción una y otra vez, un número específico de veces.

````python
for i in range(3): # 0 1 2
  print("hola")

print("chau")
````

**range()** crea x números en una secuencia, comenzando desde cero.

*Ej. range(5), generará una secuencia de 5 números.*
*Siendo: 0, 1, 2, 3, 4*

Es decir, range() genera una serie de números enteros. La variable i se utiliza para iterar sobre los número. Podemos reemplazar i con cualquier otra variable de nuestra elección.

````python
for c in range(10):
  print(c)
````

El código que se repite en el bucle for debe tener **sangría**. La sangría, son los espacios al principio de las líneas.

El código que no está correctamente sangrado dará como resultado **errores**.

A Python no le importa si se usa 2 espacios o 4 espacios (o cualquier otro número de espacios), siempre y cuando sean consistente (no cambiar despúes).

````python
#Lo siguiente dará error por falta de sangria
var = 6
for i in range(var):
  print("hola")
````

La declaración inicial del bucle debe ir seguida de dos puntos. Esto indica el inicio del bloque de iteración.

##Bucle While

Los bucles while son "poderosos" porque se pueden utilizar incluso cuando no se sabe cuántas iteraciones serán necesarias.

Los bucles while, repiten el código mientras se cumple una condición.

image.png

*Por ejemplo. Un vendedor de boletos en un teatro venderá boletos repetidamente hasta que todos los asientos estén ocupados.*

````python
asientos = 5
while asientos > 0:
  print("Vende ticket")
  asientos = asientos - 1
````

La palabra clave **while** va seguida de la condición bajo la cual se repite el código.

Cuando la condición deja de ser verdadera, "salimos" del bucle while.

En el ejemplo anterior, dejará de imprimir el mensaje cuando: *asientos = 0*

No olvidar la indentación y los dos puntos.

Los bucles generalmente incluyen **contadores.** Un contador es una variable que lleva un registro del número de iteraciones.

En el ejemplo anterior, *asientos* se utiliza como contador.

Las variables contador se actualizan dentro del bucle, por lo que cambian con cada iteración. Se establece un valor inicial fuera del bucle, como punto de partida.

Con los bucles while podemos encontrarnos con lo que se conoce como **bucle infinito**. Esto ocurre cuando la condición es siempre verdadera y el código nunca deja de repetirse.

En general, se utiliza bucles for cuando ya se conoce el número de iteraciones y bucles while cuando hay una condición que debe cumplirse.

````python
#Ejemplo
for i in range(5):
  print("hola")
  print("hola otra vez")

print("chau")
````

````python
clave = "6789"
ingresa = input()
while ingresa != clave:
  ingresa = input()
print("Clave correcta")
````

##Condicionales

La selección es como un camino en el que se bifurca. Permite a los programas decidir qué camino tomar.

image.png

Las declaraciones condicionales, o delcaraciones if-else, permiten que los programas realicen diferentes acciones basadas en las condiciones.

````python
edad = 22
if edad >= 18:
  print("Precio normal")
else:
  print("Precio con descuento")
````

Las declaraciones condicionales **if** comienzan con la palabra clave if seguida de la condición y el símbolo de dos puntos.

La declaración condicional else comienza con la palabra clave **else** seguida de dos puntos.


El código indentado debajo de la línea else se ejecuta cuando la condición no se cumple.

El código que se ejecuta dentro de los bloques if y else debe estar indentado. De lo contrario, habrá errores.

Despues de que la computadora haya terminado de ejecutar la delcaración if-else, continuará ejecutando cualquier declaración siguiente en secuencia.

Se pueden construir programas que tomen decisiones más complejas si combinamos operaciones lógicas y de comparación.

````python
edad = 32
es_estudiante = False

if edad < 18 or es_estudiante:
  print("Precio con descuento")
else:
  print("Precio normal")
````

Las declaraciones condicionales permiten programar máquinas que toman decisiones.

Habrá situaciones en las que no necesitaremos la declaración else. En este caso se puede simplificar y tener el código para el bloque selección en 1 línea.

````python
edad = 16
if edad < 18:
  print("Aplicar descuento")
  print("Pago realizado")
````

También es posible utilizar la declaración **elif** (abreviatura de "else if") para verificar más condiciones si no se cumple la primera condición.

````python
edad = int(input())
if edad < 18:
  print("Descuento junior")
elif edad >= 75:
  print("Descuento senior")
else:
  print("Sin descuento")
````

Dada una lista con nombres, crear una nueva lista sin duplicados, manteniendo el orden original. Sin usar el metodo "set()"

nombres = ['Ana', 'Luis', 'Ana', 'Pedro', 'Luis', 'María']

````python
nombres = ['Ana', 'Luis', 'Ana', 'Pedro', 'Luis', 'María']
nombres_sin_duplicados = []

for nombre in nombres:
    if nombre not in nombres_sin_duplicados:
        nombres_sin_duplicados.append(nombre)

print(nombres_sin_duplicados)  # ['Ana', 'Luis', 'Pedro', 'María']
````

````python
# Usando el método set
nombres = ['Ana', 'Luis', 'Ana', 'Pedro', 'Luis', 'María']
nombres_sin_duplicados = list(set(nombres))
print(nombres_sin_duplicados)  # ['Ana', 'Luis', 'Pedro', 'María']
````

Un set (conjunto) en Python es una colección desordenada de elementos únicos.

El método set() toma cualquier estructura iterable (como una lista, tupla o string) y devuelve un conjunto con los elementos únicos.

### Ejercicio 6

Encontrar el segundo mayor número. Dada una lista de números, encontrar el segundo valor más grande sin usar "sort()".

```
valores = [10, 5, 8, 20, 15]

```

````python
valores = [10, 5, 8, 20, 15]
mayor = max(valores)
segundo_mayor = None

for valor in valores:
    if valor != mayor:
        if segundo_mayor is None or valor > segundo_mayor:
            segundo_mayor = valor

print(segundo_mayor)  # 15
````

Usar [Python Tutor](https://pythontutor.com/) para entender el flujo

````python
# Otra forma de resolverlo
valores = [10, 5, 8, 20, 15]
mayor = segundo_mayor = float('-inf')

for num in valores:
    if num > mayor:
        segundo_mayor = mayor
        mayor = num
    elif num > segundo_mayor and num != mayor:
        segundo_mayor = num

print(segundo_mayor)  # 15
````

**Asignación múltiple**

```
variable1 = variable2 = dato
```

📌 Para tipos inmutables (como números, strings, float('-inf')):
* ✅ No están ligados entre sí.
* Cada variable apunta al mismo valor, pero como ese valor no puede cambiar, no hay problema.

🧪 Ejemplo:
```
a = b = 100
a = 200
print(b)  # 👉 100
```
* ✔️ b no se ve afectado, porque cuando se hizo a = 200, simplemente a apuntó a otro valor, sin modificar b.


📌 Para tipos mutables (listas, diccionarios, objetos personalizados, etc.):
* ⚠️ Sí están ligados.
* Ambas variables comparten la misma referencia en memoria. Si se modifica una, afectará a la otra.

🧪 Ejemplo:
```
a = b = []
a.append(1)
print(b)  # 👉 [1]
```
😬 b cambió sin que lo modifiquemos directamente, porque a y b apuntan al mismo objeto lista.

La función `find()` verifica si un caracter (o un patrón de caracteres) está presente en una cadena. La función devuelve el índice (posición) del valor dado. Si el valor dado está presente varias veces, la función devolverá la primera aparición (el índice más bajo).

Se puede definir funciones que tomen cualquier número de argumentos (incluído cero o vacío). Los argumento se colocan dentro de los paréntesis ( ) después del nombre de la función.

Una función puede requerir argumentos para completar su tarea.

````python
def saludo_personal(nombre):
  print("Hola", nombre)
  print("Bienvenido al curso de vision artificial!")

saludo_personal("Pepito")
````

Una función puede tomar tantos argumentos como sea necesario para completar su tarea.

````python
def area_rectangulo(base, altura):
  area = base * altura
  print("El resultado es:", area)

area_rectangulo(5, 4)
````

Al llamar a una función se debe usar la misma cantidad de argumentos que se han definido, en el mismo orden.

El resultado de una función puede ser enviado de regreso con la declaración **return**. Esto es particularmente útil cuando se necesita seguir usando el valor del resultado en el programa.

````python
def area_triangulo(base, altura):
  area = (base * altura)/2
  return area

resultado = area_triangulo(2, 4)
print("El resultado es:", resultado)
````

Las funciones personalizadas (o definidas por el usuario) ayudan a descomponer un programa grande en segmentos pequeños para facilitar su comprensión, mantenimiento y depuración.

Una función puede devolver múltiples valores.

*La función `rect()` ayuda a una agencia inmobiliaria a calcular el area y el perimetro de una parcela rectangular de terreno.*

Toma las dos dimensiones de la parcela como argumentos.

````python
def rect(largo, ancho):
  area = largo * ancho
  perimetro = 2 * largo + 2 * ancho
  return area, perimetro #retorna dos valores

x, y = rect(50, 100) # dos variable x, y para los valores retornados
print(x, y)
````

Los valores de retorno múltiples deben separarse por comas.

Python permite que los argumento de función tengan **valores predeterminados**. Si se llama a la función sin el argumento, el argumento obtiene su valor predeterminado.

````python
def logging(usuario="Anonimo"):
  print("Bienvenido al juego", usuario)

logging() # sin argumento
logging("Vegetta777") # con argumento
````

### Ejercicio 7

Escribir una función llamada contar_pares(lista) que reciba una lista de números y devuelva cuántos números pares hay en esa lista.

*Ejemplo*
```
contar_pares([1, 2, 3, 4, 6]) → 3
```



````python
def contar_pares(lista):
  contador = 0
  for numero in lista:
    if numero % 2 == 0:
      contador += 1
  return contador

print(contar_pares([1, 2, 3, 4, 6]))
````

### Ejercicio 8

Escribir una función promedio_positivos(lista) que devuelva el promedio solo de los números mayores a 0.
Si no hay ninguno positivo, que devuelva 0.

*Ejemplo*
```
promedio_positivos([-2, 5, 8, -1]) → 6.5
```


````python
def promedio_positivos(lista):
  suma = 0
  contador = 0
  for numero in lista:
    if numero > 0:
      suma += numero
      contador += 1
  if contador == 0:
    return 0
  else:
    return suma / contador

print(promedio_positivos([-2, 5, 8, -1]))
````

````python
# Otra forma
def promedio_positivos(lista):
    positivos = [num for num in lista if num > 0] # usando list comprehension
    if len(positivos) == 0:
        return 0
    return sum(positivos) / len(positivos)

print(promedio_positivos([-2, 5, 8, -1]))  # 6.5
````

### Ejercicio 9

Escribir una función filtrar_palabras(lista, letra) que reciba una lista de palabras y una letra, y devuelva una lista con las palabras que empiezan con esa letra (sin importar si es mayúscula o minúscula).

*Ejemplo:*

```
filtrar_palabras(["perro", "gato", "pato", "pez"], "p") → ['perro', 'pato', 'pez']
```

````python
def filtrar_palabras(lista, letra):
    resultado = []
    letra = letra.lower()
    for palabra in lista:
        if palabra.lower().startswith(letra):
            resultado.append(palabra)
    return resultado

print(filtrar_palabras(["perro", "gato", "pato", "pez"], "p"))
````

El método .startswith() sirve para verificar si una cadena de texto (string) comienza con un determinado prefijo (una letra, palabra o conjunto de caracteres).

```
cadena.startswith(prefijo[, inicio[, fin]])
```
* prefijo: El texto que queremos verificar al principio de la cadena.
* inicio (opcional): Índice desde donde empezar a buscar.
* fin (opcional): Índice donde terminar la verificación.
